```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:\\Users\\admin\\Desktop\\Cognifyz_Data_Analysis_Internship\\Dataset 1.csv")
```

```python
# Task 1 - Top Cuisines

a = df["Cuisines"].dropna().str.split(", ")
b = a.explode()

print("Top 3 Cuisines")
print(b.value_counts().head(3))
```

```python
# Task 2 - City Analysis

print("\nCity with highest restaurants")
print(df["City"].value_counts().head(1))

print("\nAverage rating of each city")
print(df.groupby("City")["Aggregate rating"].mean())
```

```python
# Task 3 - Price Range Distribution

print("\nPrice range Count")
print(df["Price range"].value_counts())

df["Price range"].value_counts().plot(kind="bar")
plt.title("Price range")
plt.show()
```

```python
# Task 4 - Online Delivery

print("\nOnline Delivery Percentage")
print(df["Has Online delivery"].value_counts(normalize=True) * 100)

print("\nAverage Rating with Online Delivery")
print(df.groupby("Has Online delivery")["Aggregate rating"].mean())
```
